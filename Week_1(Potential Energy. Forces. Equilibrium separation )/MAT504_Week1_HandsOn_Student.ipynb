## Part 3 — Microscopic stiffness

Close to equilibrium,

\[
U(r)\approx U(r_0)+\frac{1}{2}k_{\mathrm{pair}}(r-r_0)^2,
\]

where

\[
k_{\mathrm{pair}}=
\left.\frac{d^2U}{dr^2}\right|_{r=r_0}.
\]

The curvature of the potential therefore provides a measure of the local stiffness of the interaction.
